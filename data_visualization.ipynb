## Plots for categorical variables

### Pie chart

```python
labels=df["Gender"].value_counts().index
sizes=df["Gender"].value_counts().values
plt.pie(sizes,
        labels=labels,
        explode=(.1,0), # separe from center
        colors=["#6673D4","#F5CF27"],
        autopct="%1.1f%%",
        pctdistance=.6,
        shadow=True,
        startangle=90,
        radius=1.2,
        counterclock=False,
        wedgeprops={"linewidth":3},
        textprops={'fontsize':12,'color':'black','fontweight':'normal',
                   'family':'Candara'},
        center=(0,0),
        frame=False,
        normalize=True)
plt.title("Gender distribution",
          fontsize=14,
          fontweight="normal", # bold
          fontfamily="Candara",
          pad=20);
```

### Bar plot

```python
categories=df["Gender"].value_counts().index
values=df["Gender"].value_counts().values
xticks_dict={"M":"Masculine","F":"Feminine"}
labels=[xticks_dict[i] for i in categories]
```

```python
plt.bar(categories,values,
        color=["#6673D4","#F5CF27"],
        edgecolor="black",
        linewidth=.8,
        width=.6) # width and separation between bars
plt.xlabel("Gender",fontfamily="Candara",fontsize=12)
plt.ylabel("Count",fontfamily="Candara",fontsize=12)
plt.xticks(range(len(values)),labels,fontfamily="Candara",fontsize=12)
plt.yticks(fontfamily="Candara",fontsize=12)
plt.grid(axis="y",linestyle="--",alpha=.4) # Horizontal grid
ax=plt.gca() # Axes object of the current plot
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.title("Gender distribution",
          fontsize=16,
          fontweight="normal",
          fontfamily="Candara",
          pad=5);
```

### Lollipop chart

Useful when you have many categories to plot. Elegant way to represent categorical variables with many categories

```python
categories=df["Gender"].value_counts().index
values=df["Gender"].value_counts().values
xticks_dict={"M":"Masculine","F":"Feminine"}
labels=[xticks_dict[i] for i in categories]
trim_margin=.3
stem_space=.6
x=np.arange(len(values))*stem_space
plt.figure(figsize=(5,4))
markerline,stemlines,baseline=plt.stem(x,values,basefmt=" ")
plt.xticks(x,labels,fontfamily="Candara",fontsize=12)
plt.xlim(min(x)-trim_margin,max(x)+trim_margin)
plt.ylabel("Count",fontfamily="Candara")
plt.title("Gender distribution",
          fontsize=16,
          fontweight="normal",
          fontfamily="Candara",
          pad=5);
```

```python
markerline, stemlines, baseline = plt.stem(x, values, basefmt=" ")

plt.setp(markerline, markersize=8)
plt.setp(stemlines, linewidth=1.5)
```
