## Plots for categorical variables

### Pie chart

```python
labels=df["Gender"].value_counts().index
sizes=df["Gender"].value_counts().values
plt.pie(sizes,
        labels=labels,
        explode=(.1,0), # separe from center
        colors=["#6673D4","#F5CF27"],
        autopct="%1.1f%%",
        pctdistance=.6,
        shadow=True,
        startangle=90,
        radius=1.2,
        counterclock=False,
        wedgeprops={"linewidth":3},
        textprops={'fontsize':12,'color':'black','fontweight':'normal',
                   'family':'Candara'},
        center=(0,0),
        frame=False,
        normalize=True);
```
